```python
import pandas as pd
import os
from pathlib import Path
```

```python
PROJECT_ROOT = Path().resolve().parent
DATA_PATH = PROJECT_ROOT / "data"
file_path = DATA_PATH/"64fa2910-4d48-469d-92d8-d00086d1e462_2018_2019.csv"

df = pd.read_csv(file_path)
pd.set_option('display.max_columns', 50)
```

```python
print(df.columns)
```

```python
date_columns = []
for col in df.columns:
    if 'date' in col:
        date_columns.append(col)
        print(f"COL:{col}")
        print(df[col].head(5))
```

Turining str dataes to date time

```python
df = pd.read_csv(
    filepath_or_buffer=file_path,
    parse_dates=date_columns,
    dayfirst=True 
)
#uing time as 9:00:00 eveyone elese uses 09:00:00
df["tender_datepublished"] = pd.to_datetime(df["tender_datepublished"], errors="coerce",dayfirst=True)
```
